```python
from openai import OpenAI


system_prompt = "I want you to be a career adivsor who can help me find the best career for me based on my skills and interests."
user_prompt = "I am trying to start a new career in software development. What are the best skills and interests I should focus on?"


messages = [
    {"role": "system", "content": system_prompt},
    {"role": "user", "content": user_prompt}
]

openai = OpenAI()
response = openai.chat.completions.create(
    model="gpt-5-nano",
    messages=messages
)

print(response.choices[0].message.content)
```
