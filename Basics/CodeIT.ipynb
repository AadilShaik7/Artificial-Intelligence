```python
import gradio as gr
 
with gr.Blocks(title="Code It") as main:
    gr.Markdown("#  Code It")
    
    with gr.Row():

        with gr.Column(scale=1):
            language_dropdown = gr.Dropdown(
                choices=["Python", "Java", "C++", "JavaScript", "Go", "TypeScript", "Auto-detect"],
                label="Programming Language",
                value="Python",
                interactive=True
            )
            
            code_input = gr.Textbox(
                label="Paste Your Code",
                lines=20,
                placeholder="def hello(name):\n    return f'Hello, {name}!'",
                max_lines=30
            )
            
            gr.Markdown("### Select Analysis")
            
            checkbox_explain = gr.Checkbox(label="Explain Code", value=True)
            checkbox_bugs = gr.Checkbox(label="Find Bugs", value=True)
            checkbox_optimize = gr.Checkbox(label="Optimize Code", value=True)
            
            analyze_btn = gr.Button("Analyze Code", variant="primary", size="lg")
        

        with gr.Column(scale=1):
            with gr.Tabs():
                with gr.Tab("Explanation"):
                    explain_output = gr.Textbox(
                        label="Output",
                        lines=20,
                        max_lines=30,
                        interactive=False
                    )
                
                with gr.Tab("Bugs"):
                    bugs_output = gr.Textbox(
                        label="Output",
                        lines=20,
                        max_lines=30,
                        interactive=False
                    )
                
                with gr.Tab("Optimization"):
                    optimize_output = gr.Textbox(
                        label="Output",
                        lines=20,
                        max_lines=30,
                        interactive=False
                    )
    main.launch()
```
