```python
# Get Llamafrom openai import OpenAI
from openai import OpenAI
client = OpenAI(
    base_url="http://localhost:11434/v1",
    api_key="ollama"
)

System_Prompt_Professor = """
Your name is Raj. You are a polite physics professor.
You answer physics questions clearly but can become slightly sarcastic
if the student keeps bothering you.
"""

System_Prompt_Student = """
You are a curious physics student.
Your job is to keep asking physics questions and slightly annoy the professor.
"""

prof_messages = [{"role": "system", "content": System_Prompt_Professor}]
student_messages = [{"role": "system", "content": System_Prompt_Student}]

msg = input("Initial student message: ")

prof_messages.append({
    "role": "user",
    "content": msg
})

turns = 3

for i in range(turns):

    print("\n Professor:", end=" ", flush=True)

    stream = client.chat.completions.create(
        model="llama3.2",
        messages=prof_messages,
        stream=True
    )

    prof_reply = ""

    for chunk in stream:
        if chunk.choices[0].delta.content:
            token = chunk.choices[0].delta.content
            print(token, end="", flush=True)
            prof_reply += token

    print()

    prof_messages.append({"role": "assistant", "content": prof_reply})
    student_messages.append({"role": "user", "content": prof_reply})


    print("\n Student:", end=" ", flush=True)

    stream = client.chat.completions.create(
        model="llama3.2",
        messages=student_messages,
        stream=True
    )

    student_reply = ""

    for chunk in stream:
        if chunk.choices[0].delta.content:
            token = chunk.choices[0].delta.content
            print(token, end="", flush=True)
            student_reply += token

    print()

    student_messages.append({"role": "assistant", "content": student_reply})
    prof_messages.append({"role": "user", "content": student_reply})
```
